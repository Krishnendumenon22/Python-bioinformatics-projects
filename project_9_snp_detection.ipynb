# Project 9: SNP Detection

A single nucleotide polymorphism (SNP) is a position where a patient's DNA base differs from the reference sequence. This notebook compares a patient sequence with a reference sequence and reports every position where the base differs.

**Assumptions and limits:** the function expects two sequences that are already aligned and of equal length. It detects substitutions only, so insertions and deletions are not handled.

## The function

```python
def find_snps(patient, reference):
    """Compare a patient sequence with a reference sequence and report SNPs.

    Both sequences must be the same length (already aligned).
    Returns a list of (position, reference_base, patient_base) tuples.
    """
    if len(patient) != len(reference):
        raise ValueError("Sequences must be the same length (aligned).")

    snps = []
    for i, (p, r) in enumerate(zip(patient.upper(), reference.upper()), start=1):
        if p != r:
            snps.append((i, r, p))
            print(f"SNP at position {i}: {r} (reference) -> {p} (patient)")
    return snps
```

## Example
The positions are 1-based, and each SNP is reported as reference base -> patient base.

```python
patient = "ATGGGATACAGATACGTGAGT"
reference = "ATGGGAGACAGATACGTGACT"

snps = find_snps(patient, reference)
print("Total SNPs found:", len(snps))
```
